```julia
function get_phot_err(timing_err,rp,τ,exp_time)
    """
    in limit where σ_τ << τ and exposure time << τ
    phot_err ∝ √(Ṅτ) , for count rate of star Ṅ
    δ = (rp/rstar)^2
    ΔN = σ_τ δ Ṅ , count deficit
    √(Ṅτ) = σ_τ δ Ṅ
    σ_τ = √(τ/Ṅ) × δ^(-1)
    σ_t = σ_τ /√2
    """
    rstar=1
    δ = (rp/rstar)^2
    phot_err =(δ * timing_err * sqrt(2))/ sqrt(τ * exp_time)
    return phot_err
end
phot_err=get_phot_err(0.00034,0.0091705,(400/24/60/60),(2/24/60))
```

```julia
function get_time_error(phot_err,rp,τ,exp_time)
    rstar=1
    δ = (rp/rstar)^2
   return phot_err*sqrt(τ * exp_time)/(δ*sqrt(2))
end
get_time_error(34/1000000,0.0091705,(400/24/60/60),22)
```
